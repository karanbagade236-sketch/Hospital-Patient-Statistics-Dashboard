```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Read external dataset
df = pd.read_csv("hospital_patient_statistics.csv")

# Convert Month column into date format
df["Month"] = pd.to_datetime(df["Month"])

# Create subplots
fig, (axes1, axes2) = plt.subplots(
    1, 2, figsize=(14, 5)
)

# Monthly patient admissions
axes1.plot(
    df["Month"],
    df["Patients Admitted"],
    marker="^",
    color="red"
)

axes1.set_title("Monthly Patient Admissions")
axes1.set_xlabel("Month")
axes1.set_ylabel("Patients Admitted")

# Emergency cases
axes2.bar(
    df["Month"],
    df["Emergency Cases"],
    width=20,
    color="green"
)

axes2.set_title("Emergency Cases by Month")
axes2.set_xlabel("Month")
axes2.set_ylabel("Emergency Cases")

# Adjust layout
plt.tight_layout()

# Save output
plt.savefig("hospital_patient_statistics.png", dpi=300)

# Display graph
plt.show()
```
